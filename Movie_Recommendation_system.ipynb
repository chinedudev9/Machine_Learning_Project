Import Dependencies

```python
import pandas as pd
import numpy as np
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies_data = pd.read_csv('/content/movies.csv', engine='python')
```

```python
movies_data.head()
```

```python
#checking for null data
movies_data.isnull().sum()
```

```python
#selecting the relevant column
selected_data = ['genres', 'keywords', 'tagline', 'cast', 'director']
print(selected_data)
```

```python
#replacing the missing data in the selected_data with empty string by looping
for i in selected_data:
  movies_data[i] = movies_data[i].fillna('')
```

```python
#concatenate the data in the selected_data
combined_data = movies_data['genres']+' '+movies_data['keywords']+' '+movies_data['tagline']+' '+movies_data['cast']+' '+movies_data['director']
```

```python
print(combined_data)
```

```python
combined_data.isnull().sum()
```

```python
#convert the text data to numeriacal data
vectorizer = TfidfVectorizer()
```

```python
numerical_data = vectorizer.fit_transform(combined_data)
```

```python
print(numerical_data)
```

Finding the similarity

```python
cosine_similarity = cosine_similarity(numerical_data)
```

```python
print(cosine_similarity)
```

```python
cosine_similarity.shape
```

```python
#get movie from user
movie_name = input('Enter the movie name: ')
```

```python
#get the list of all the title
title = movies_data['title'].tolist()
print(title)
```

```python
#recommending the right movie
find_close_match = difflib.get_close_matches(movie_name, title)
print(find_close_match)
```

```python
first_movie = find_close_match[0]
print(first_movie)
```

```python
#find the similarity score
index_of_movie = movies_data[movies_data.title == first_movie]['index'].values[0]
print(index_of_movie)
```

```python
#getting similar movies
similarity_score = list(enumerate(cosine_similarity[index_of_movie]))
print(similarity_score)
```

```python
len(similarity_score)
```

```python
# sorting the similarity in reverse order
sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse = True)
print(sorted_similar_movies)
```

```python
# loop through and print similar movies
print('You will Also like similar movies like that of', movie_name, '\n')

i = 1

for movie in sorted_similar_movies:
  index = movie[0]
  title_index = movies_data[movies_data.index == index] ['title'].values[0]
  if(i < 21):
    print(i, '.', title_index)
    i+=1
```

The Recommendation System

```python
# Recommendation system
movie_name = input('Enter the movie name: ')
title = movies_data['title'].tolist()
find_close_match = difflib.get_close_matches(movie_name, title)

if len(find_close_match) == 0:
    print('Sorry, no close match found for the movie name you entered.')
else:
    first_movie = find_close_match[0]
    index_of_movie = movies_data[movies_data.title == first_movie]['index'].values[0]
    similarity_score = list(enumerate(cosine_similarity[index_of_movie]))
    sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse = True)

    print('This Movies are similar to', first_movie, '\n')

    i = 1

    for movie in sorted_similar_movies:
      index = movie[0]
      title_index = movies_data[movies_data.index == index] ['title'].values[0]
      if(i < 21):
        print(i, '.', title_index)
        i+=1
```
